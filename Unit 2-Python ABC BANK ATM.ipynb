```python
# IDFC ATM
user = input("Please enter your name and card details: ")

print("Welcome to IDFC Bank!")
print(f"You are {user.capitalize()}. These are your IDFC Bank card details.")
print("Please verify your details.")

dt = input("Are your details correct? Press Yes or No: ").strip().lower()

if dt != "yes":
    print("Please try again to verify your details.")
else:
    print("Thank you for verifying your details.")

    services = [
        "Withdraw Money: To withdraw cash from your account.",
        "Balance Inquiry: To check your account balance.",
        "Mini Statement: To view a summary of your transactions.",
        "Deposit Money: To deposit cash into your account.",
        "Bill Pay: To pay bills directly from your account.",
        "Transfer Funds: To transfer money to other accounts.",
        "Other",
    ]

    print("\nPlease select a service option:")
    for i, service in enumerate(services, start=1):
        print(f"{i}. {service}")

    choice = input("Enter the option number: ").strip()

    if choice.isdigit():
        choice = int(choice)

        if 1 <= choice <= len(services):
            print(f"You selected: {services[choice - 1]}")
        else:
            print("That option is not on the list.")
    else:
        print("Please enter a number.")
```
